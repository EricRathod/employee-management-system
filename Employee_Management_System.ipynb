Creating Database

```python
import sqlite3

conn = sqlite3.connect("employee_management.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS employees(
    employee_id INTEGER PRIMARY KEY AUTOINCREMENT,
    full_name TEXT,
    department TEXT,
    job_title TEXT,
    email TEXT,
    phone TEXT,
    salary REAL,
    joining_date TEXT
)
""")

conn.commit()
conn.close()

print("Database Created Successfully")
```

Search Employee

```python
import sqlite3
import pandas as pd

search_name = input("Enter employee name to search: ")

conn = sqlite3.connect("employee_management.db")

query = """
SELECT * FROM employees
WHERE full_name LIKE ?
"""

df = pd.read_sql_query(query, conn, params=('%' + search_name + '%',))

conn.close()

df
```

Update Employee

```python
import sqlite3

employee_id = int(input("Enter Employee ID to update: "))
new_department = input("Enter new department: ")
new_job_title = input("Enter new job title: ")
new_salary = float(input("Enter new salary: "))

conn = sqlite3.connect("employee_management.db")
cursor = conn.cursor()

cursor.execute("""
UPDATE employees
SET department = ?, job_title = ?, salary = ?
WHERE employee_id = ?
""", (new_department, new_job_title, new_salary, employee_id))

conn.commit()
conn.close()

print("Employee Updated Successfully")
```

Delete Employee

```python
import sqlite3

employee_id = int(input("Enter Employee ID to delete: "))

conn = sqlite3.connect("employee_management.db")
cursor = conn.cursor()

cursor.execute("""
DELETE FROM employees
WHERE employee_id = ?
""", (employee_id,))

conn.commit()
conn.close()

print("Employee Deleted Successfully")
```

Department Report

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("employee_management.db")

query = """
SELECT
department,
COUNT(*) AS Total_Employees,
ROUND(AVG(salary),2) AS Average_Salary
FROM employees
GROUP BY department
"""

report = pd.read_sql_query(query, conn)

conn.close()

report
```
